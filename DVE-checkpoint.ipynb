#### 1.2 Numerical Fields

```python
# gallons, miles, and odometer
gallons_missing_perc = round(df["gallons"].isna().sum() / df["gallons"].count(), 3) * 100
miles_missing_perc = round(df["miles"].isna().sum() / df["miles"].count(), 3) * 100
odometer_missing_perc = round(df["odometer"].isna().sum() / df["odometer"].count(), 3) * 100

print("Gallons missing percentage \t:", gallons_missing_perc)
print("Miles missing percentage \t:", miles_missing_perc)
print("Odometer missing percentage \t:", odometer_missing_perc)
```

```python
df_filtered = df[df[["miles", "gallons", "mpg"]].notna().all(axis=1)]
df_filtered[["miles", "gallons", "mpg"]]
# df_filtered = df[df[["col1", "col2", "col3"]].notna().all(axis=1)]
```

```python
#before
df[["miles", "gallons", "mpg"]].isna().sum()
```

```python
# miles = gallons * mpg
# gallons = miles / mpg
# mpg = miles / gallons

cols = ["miles", "gallons", "mpg"]

#Check for decimals
for col in cols:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(",", ""),
        errors="coerce"
    )

#miles
df.loc[
    df["miles"].isna() & df["gallons"].notna() & df["mpg"].notna(),
    "miles"
] = (
    df["gallons"].astype(float) * df["mpg"].astype(float)
)

#gallons
df.loc[
    df["gallons"].isna() & df["miles"].notna() & df["mpg"].notna() & (df["mpg"] != 0),
    "gallons"
] = (
    df["miles"].astype(float) / df["mpg"].astype(float)
)

#mpg
df.loc[
    df["mpg"].isna() & df["miles"].notna() & df["gallons"].notna() & (df["gallons"] != 0),
    "mpg"
] = (
    df["miles"].astype(float) / df["gallons"].astype(float)
)
```

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 12))

columns = ["miles", "gallons", "mpg"]

for ax, col in zip(axes, columns):
    ax.hist(df[col].dropna(), bins=30)
    ax.set_title(f"Distribution of {col.title()}")
    ax.set_xlabel(col.title())
    ax.set_ylabel("Frequency")
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
columns = ["miles", "gallons", "mpg"]

for col in columns:
    plt.figure(figsize=(10, 5))
    
    df[col].dropna().plot(
        kind="hist",
        bins=30,
        density=True,
        alpha=0.6
    )
    
    df[col].dropna().plot(
        kind="kde"
    )
    
    plt.title(f"Distribution of {col.title()}")
    plt.xlabel(col.title())
    plt.ylabel("Density")
    plt.grid(alpha=0.3)
    plt.show()
```
