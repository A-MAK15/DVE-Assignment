```python
# gallons, miles, and odometer
#Q : Identify what percentage of gallons, miles, and odometer entries are missing
gallons_missing_perc = round(df["gallons"].isna().sum() / (df["gallons"].count() + df["gallons"].isna().sum()), 3) * 100
miles_missing_perc = round(df["miles"].isna().sum() / (df["miles"].count() + df["miles"].isna().sum()), 3) * 100
odometer_missing_perc = round(df["odometer"].isna().sum() / (df["odometer"].count() + df["odometer"].isna().sum()), 3) * 100

print("Gallons missing percentage \t:", gallons_missing_perc)
print("Miles missing percentage \t:", miles_missing_perc)
print("Odometer missing percentage \t:", odometer_missing_perc)
```

```python
df_filtered = df[df[["miles", "gallons", "mpg"]].notna().all(axis=1)]
df_filtered[["miles", "gallons", "mpg"]]
# df_filtered = df[df[["col1", "col2", "col3"]].notna().all(axis=1)]
```

```python
#before
df[["miles", "gallons", "mpg"]].isna().sum()
```

```python
# miles = gallons * mpg
# gallons = miles / mpg
# mpg = miles / gallons

cols = ["miles", "gallons", "mpg"]

#Check for decimals
for col in cols:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(",", ""),
        errors="coerce"
    )

#miles
df.loc[
    df["miles"].isna() & df["gallons"].notna() & df["mpg"].notna(),
    "miles"
] = (
    df["gallons"].astype(float) * df["mpg"].astype(float)
)

#gallons
df.loc[
    df["gallons"].isna() & df["miles"].notna() & df["mpg"].notna() & (df["mpg"] != 0),
    "gallons"
] = (
    df["miles"].astype(float) / df["mpg"].astype(float)
)

#mpg
df.loc[
    df["mpg"].isna() & df["miles"].notna() & df["gallons"].notna() & (df["gallons"] != 0),
    "mpg"
] = (
    df["miles"].astype(float) / df["gallons"].astype(float)
)
```

```python
parts = df["user_url"].str.rstrip("/").str.split("/")
# print(parts[0][0])

df["make"] = None
df["model"] = None
df["year"] = None
df["user_id"] = None

# # Standard URLs (7 parts)
mask7 = parts.str.len() == 7
df.loc[mask7, "make"] = parts[mask7].str[3]
df.loc[mask7, "model"] = parts[mask7].str[4]

#Missing model (6 parts)
mask6 = parts.str.len() == 6
df.loc[mask6, "make"] = parts[mask6].str[3]
df.loc[mask6, "model"] = parts[mask6].str[4]

#Category URLs (8 parts)
mask8 = parts.str.len() == 8
df.loc[mask8, "make"] = parts[mask8].str[4]
df.loc[mask8, "model"] = parts[mask8].str[5]
```

```python
# Year (optional 4 digits)
df["year"] = pd.to_numeric(
    df["user_url"].str.extract(r'/(\d{4})/(?=\d+$)')[0],
    errors="coerce"
)

# User ID (last number in the URL)
df["user_id"] = pd.to_numeric(
    df["user_url"].str.extract(r'(\d+)$')[0]
)
```
